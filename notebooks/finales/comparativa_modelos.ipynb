## Modelos Supervisados
Son modelos que se entrenan con datos etiquetados, es decir, cada instancia de entrenamiento tiene una etiqueta o resultado asociado. El objetivo es aprender una función que pueda predecir la etiqueta de nuevas instancias basándose en las características de los datos. Usa tanto la información de entrada como la respuesta o resultado correcto esperado. El objetivo es que el modelo aprenda estos patrones para realizar predicciones sobre datos nuevos.

### Random forest

Random forest es un modelo supervisado de aprendizaje automático que se utiliza para clasificación y regresión. Se basa en la construcción de múltiples árboles de decisión durante el entrenamiento y la salida de la clase que es el modo de las clases (clasificación) o la media de las predicciones (regresión) de los árboles individuales.

### Neural network
Una red neuronal es un modelo de aprendizaje automático inspirado en la estructura del cerebro humano. Consiste en capas de nodos (neuronas) que procesan información y aprenden a realizar tareas específicas a través del entrenamiento con datos.

### Decision tree
Un árbol de decisión es un modelo de aprendizaje automático que se utiliza para clasificación y regresión. Se representa como un árbol donde cada nodo interno representa una prueba en una característica, cada rama representa el resultado de la prueba y cada hoja representa una clase o valor de predicción.

### Gradient boosting
Gradient boosting es un modelo de aprendizaje automático que se utiliza para clasificación y regresión. Se basa en la construcción de modelos secuenciales, donde cada modelo intenta corregir los errores del modelo anterior. El resultado final es una combinación de todos los modelos individuales.

### Logistic regression
La regresión logística es un modelo de aprendizaje automático utilizado para clasificación binaria. Se basa en la función logística para modelar la probabilidad de que una instancia pertenezca a una clase específica.
